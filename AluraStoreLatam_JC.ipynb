## Visualizacion

```python
df_datos_tiendas = pd.DataFrame(
    {
        "Tiendas": ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"],
        "Ingreso Total": [
            tienda.Precio.sum(),
            tienda2.Precio.sum(),
            tienda3.Precio.sum(),
            tienda4.Precio.sum()
        ],
        "Costo de Envio": [
            tienda["Costo de envío"].mean(),
            tienda2["Costo de envío"].mean(),
            tienda3["Costo de envío"].mean(),
            tienda4["Costo de envío"].mean()
        ],
        "Calificacion": [
            tienda.Calificación.mean(),
            tienda2.Calificación.mean(),
            tienda3.Calificación.mean(),
            tienda4.Calificación.mean()
        ],

    }

)
df_datos_tiendas
```

## Ingreso total por tienda

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12, 6))
df_datos_tiendas.plot(kind="bar", x="Tiendas", y="Ingreso Total", color="purple", xlabel="Tiendas", ylabel="Ingreso Total", title="Ingreso Total por tienda", rot=0, ax=ax)

grafico_ingresos_totales = fig
```

## Extra

  ### Utilice los datos de latitud (lat) y longitud (lon) para mapear las ventas de cada tienda y analizar la distribución geográfica de los productos vendidos.

```python
tienda
```

```python
tienda.lat.unique()
```

```python
tienda.lon.unique()
```

```python
df_coordenadas_tienda1 = tienda.groupby(["lat", "lon"]).size().to_frame(name="Ventas").sort_values("Ventas").sort_values(by="Ventas").reset_index()
df_coordenadas_tienda1["Tienda"] = "tienda 1"

df_coordenadas_tienda1 = df_coordenadas_tienda1[["Tienda", "lon", "lat", "Ventas"]]
df_coordenadas_tienda1
```

```python
df_coordenadas_tienda2 = tienda2.groupby(["lat", "lon"]).size().to_frame(name="Ventas").sort_values("Ventas").sort_values(by="Ventas").reset_index()
df_coordenadas_tienda2["Tienda"] = "tienda 2"
df_coordenadas_tienda2 = df_coordenadas_tienda2[["Tienda", "lon", "lat", "Ventas"]]
df_coordenadas_tienda2
```

```python
df_coordenadas_tienda3 = tienda3.groupby(["lat", "lon"]).size().to_frame(name="Ventas").sort_values("Ventas").sort_values(by="Ventas").reset_index()
df_coordenadas_tienda3["Tienda"] = "tienda 3"
df_coordenadas_tienda3 = df_coordenadas_tienda3[["Tienda", "lon", "lat", "Ventas"]]
df_coordenadas_tienda3
```

```python
df_coordenadas_tienda4 = tienda4.groupby(["lat", "lon"]).size().to_frame(name="Ventas").sort_values("Ventas").sort_values(by="Ventas").reset_index()
df_coordenadas_tienda4["Tienda"] = "tienda 4"
df_coordenadas_tienda4 = df_coordenadas_tienda4[["Tienda", "lon", "lat", "Ventas"]]
df_coordenadas_tienda4
```

```python
#unir 4 dataframes filtrados por coordenada
lista_tiendas = [df_coordenadas_tienda1, df_coordenadas_tienda2, df_coordenadas_tienda3, df_coordenadas_tienda4]

df_grupal = pd.concat(lista_tiendas, ignore_index=True)
df_grupal
```

### Genere gráficos de dispersión o mapas de calor (Heatmaps) para visualizar datos e identificar áreas con la mayor concentración de ventas.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))
sns.scatterplot(
    data=df_grupal,
    x='lon',
    y='lat',
    hue='Tienda',    # Esto pondrá un color distinto a cada una de las 4 tiendas
    size='Ventas',    # El tamaño del punto será según las ventas
    palette='viridis'
)

plt.title('Distribución Geográfica de Ventas - 4 Tiendas')
plt.show()
```

### Explore si algunas tiendas tienen un rendimiento superior o inferior al esperado en determinadas regiones e identifique si existen patrones geográficos que puedan influir en el rendimiento de las tiendas.


- Tienda 1: Presenta un rendimiento acorde a lo esperado, ejerciendo un dominio claro en la zona norte (entre longitud $-76$ y $-74$, latitud $10$ a $12$). Es la tienda con el alcance geográfico más amplio de todo el grupo.

- Tienda 2: Muestra datos más dispersos y escasos. Sin embargo, al ser la segunda con mayores ingresos, se concluye que su modelo de negocio se basa en la venta de productos de alto valor unitario (calidad/precio) en lugar de volumen (cantidad).

- Tienda 3: Se identifica como la opción preferida por los clientes en la zona central (coordenadas $-74, 4$), superando el rendimiento de la Tienda 2 en esa ubicación específica.

- Tienda 4: Refleja un rendimiento óptimo y concentrado en una zona geográfica específica (identificada por sus 3 círculos agrupados). Esto sugiere una fuerte lealtad y confianza de los clientes en ese mercado local.
